# بررسی محیط کاری پروژه تشخیص صدا

این نوت‌بوک نسخه پایتون، مسیر اجرایی آن و نسخه کتابخانه‌های مورد استفاده را در زمان اجرای نوت‌بوک نمایش می‌دهد. سلول زیر را در همان محیطی اجرا کنید که پروژه در آن توسعه داده می‌شود.

```python
import sys
import platform
import importlib.metadata as metadata

print('Python version:', sys.version)
print('Python executable:', sys.executable)
print('Platform:', platform.platform())

packages = ['numpy', 'matplotlib', 'scipy', ' librosa', 'soundfile', 'scikit-learn', 'ipykernel', 'jupyter']
for package in packages:
    package = package.strip()
    try:
        print(f'{package}: {metadata.version(package)}')
    except metadata.PackageNotFoundError:
        print(f'{package}: not installed')
```

## نکته
نسخه‌ها عمداً هنگام اجرا خوانده می‌شوند تا گزارش با محیط واقعی شما هماهنگ باشد. اگر کتابخانه‌ای نصب نیست، در خروجی با عبارت `not installed` مشخص می‌شود. برای نصب کتابخانه‌های موردنیاز، در ترمینال همان محیط می‌توانید از دستور `python -m pip install numpy matplotlib scipy librosa soundfile scikit-learn jupyter ipykernel` استفاده کنید.
